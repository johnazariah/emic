# Chapter 2: The Quantum Twist - Code Examples

This notebook demonstrates **interference** — the key feature that makes quantum mechanics quantum.

Key insight: **Amplitudes are complex, and phases matter.**

```python
import numpy as np

np.set_printoptions(precision=3, suppress=True)
```

## Destructive Interference

Two paths with opposite phases **cancel completely**.

```python
# Two amplitudes with different phases
alpha_A = 1 / np.sqrt(2) * np.exp(1j * 0)  # phase = 0
alpha_B = 1 / np.sqrt(2) * np.exp(1j * np.pi)  # phase = π (opposite)

# Add amplitudes (quantum)
alpha_total = alpha_A + alpha_B
prob_quantum = np.abs(alpha_total) ** 2

# Add probabilities (classical)
prob_classical = np.abs(alpha_A) ** 2 + np.abs(alpha_B) ** 2

print(f"Amplitude A: {alpha_A:.3f} (phase = 0)")
print(f"Amplitude B: {alpha_B:.3f} (phase = π)")
print(f"Total amplitude: {alpha_total:.3f}")
print()
print(f"Quantum probability: {prob_quantum:.3f}")
print(f"Classical probability: {prob_classical:.3f}")
print()
print("🔥 Destructive interference: quantum probability is ZERO!")
```

## Constructive Interference

When phases align, amplitudes **add up**.

```python
# Two amplitudes with the same phase
alpha_A = 1 / np.sqrt(2) * np.exp(1j * 0)  # phase = 0
alpha_B = 1 / np.sqrt(2) * np.exp(1j * 0)  # phase = 0 (same)

alpha_total = alpha_A + alpha_B
prob_quantum = np.abs(alpha_total) ** 2

print(f"Amplitude A: {alpha_A:.3f}")
print(f"Amplitude B: {alpha_B:.3f}")
print(f"Total amplitude: {alpha_total:.3f}")
print(f"Quantum probability: {prob_quantum:.3f}")
print()
print("✨ Constructive interference: probability is enhanced!")
```

## Interference Pattern

Let's see how probability varies with phase difference.

```python
# Phase difference sweep
phases = np.linspace(0, 2 * np.pi, 9)

print("Phase Δφ    | cos(Δφ) | P_quantum | Interpretation")
print("-" * 55)

for phi in phases:
    alpha_A = 1 / np.sqrt(2)
    alpha_B = 1 / np.sqrt(2) * np.exp(1j * phi)

    alpha_total = alpha_A + alpha_B
    prob = np.abs(alpha_total) ** 2

    if np.abs(np.cos(phi) - 1) < 0.01:
        interp = "Constructive"
    elif np.abs(np.cos(phi) + 1) < 0.01:
        interp = "Destructive"
    else:
        interp = "Partial"

    print(f"{phi:6.2f} rad | {np.cos(phi):+6.2f}  | {prob:9.3f} | {interp}")
```

## Phase in the Density Matrix

The off-diagonal elements carry phase information.

```python
def density_matrix_with_phase(phi):
    """Density matrix for (|0⟩ + e^{iφ}|1⟩)/√2"""
    psi = np.array([1, np.exp(1j * phi)]) / np.sqrt(2)
    return np.outer(psi, np.conj(psi))


print("Density matrices for |ψ⟩ = (|0⟩ + e^{iφ}|1⟩)/√2")
print("=" * 50)

for phi, name in [(0, "φ=0"), (np.pi / 2, "φ=π/2"), (np.pi, "φ=π")]:
    rho = density_matrix_with_phase(phi)
    print(f"\n{name}:")
    print(rho)
    print(f"Diagonal (probabilities): [{rho[0, 0].real:.2f}, {rho[1, 1].real:.2f}]")
    print(f"Off-diagonal: {rho[0, 1]:.3f}")
```

## Key Observation

Notice that:
- **Diagonals stay the same** (both are 0.5) — same measurement probabilities
- **Off-diagonals change** — they encode the phase

The phase is "hidden" from simple measurements but affects interference experiments.

## Complex Numbers: Quick Visualization

```python
# Complex numbers in polar form
magnitudes = [1, 1, 1, 1]
phases = [0, np.pi / 2, np.pi, 3 * np.pi / 2]
names = ["0", "π/2", "π", "3π/2"]

print("Complex numbers with unit magnitude, different phases:")
print("=" * 50)
for mag, phase, name in zip(magnitudes, phases, names):
    z = mag * np.exp(1j * phase)
    print(f"e^(i·{name:4s}) = {z.real:+.3f} {z.imag:+.3f}i")

print("\nAdding opposite phases:")
z1 = np.exp(1j * 0)
z2 = np.exp(1j * np.pi)
print(f"e^(i·0) + e^(i·π) = {z1:.1f} + {z2:.1f} = {z1 + z2:.1f}")
print("They cancel!")
```

## Key Takeaway

> **Amplitudes are complex. Phases cause interference.**
>
> - Same phase → constructive interference → probability enhanced
> - Opposite phase → destructive interference → probability reduced (or zero!)
>
> This is encoded in the **off-diagonal elements** of the density matrix.
> Classical systems (diagonal only) can't interfere.
